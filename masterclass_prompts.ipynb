## De Junior a Senior: El Arte de Guiar a la Inteligencia Artificial

### ¿Por qué necesitamos estructurar nuestros prompts?
Un modelo de IA es como un desarrollador Junior sumamente rápido, pero sin contexto de tu negocio. Si le das instrucciones vagas (ej. *"hazme una app de ventas"*), te dará código espagueti o muy genérico. Si le das instrucciones estructuradas, te entregará software de grado de producción.

Para lograr esto, utilizaremos **Etiquetas XML**. Estas ayudan al agente a entender perfectamente su propósito sin mezclar conceptos:
- `<role>`: Quién es la IA en este contexto (ej. Arquitecto Cloud, Desarrollador Frontend Senior).
- `<context>`: El escenario o problema de negocio de la vida real que estamos resolviendo.
- `<instruction>`: El paso a paso técnico y detallado de lo que necesitamos.
- `<format>`: Cómo queremos que nos entregue la respuesta (ej. estructura de archivos, un solo código ejecutable, etc.).

## 🐍 1. Python + Tkinter: Sistema de Gestión de Inventario

**Caso de uso de la vida real:** Una pequeña refaccionaria automotriz necesita un sistema de escritorio ligero para registrar entradas y salidas de piezas.
**Herramienta:** Python con Tkinter (interfaz gráfica nativa, sin requerir servidor).

```xml
<role>
Eres un Arquitecto de Software y Desarrollador Python Senior, experto en la construcción de interfaces gráficas de escritorio robustas usando Tkinter, con un enfoque estricto en Clean Code, principios SOLID y arquitecturas MVC.
</role>

<context>
Una refaccionaria automotriz local con alto volumen de operaciones diarias necesita digitalizar su inventario de manera urgente. Actualmente operan con papel y lápiz, lo que causa pérdidas y descontrol. Requieren un sistema local ligero, que no dependa de internet ni instalaciones complejas de bases de datos. Los usuarios finales (personal de almacén) tienen conocimientos técnicos nulos, por lo que la interfaz debe ser a prueba de errores, intuitiva y prevenir pérdida de datos por acciones accidentales.
</context>

<instruction>
Desarrolla una aplicación de escritorio completa usando Python y Tkinter para la gestión integral de inventario.
La aplicación debe incluir lo siguiente:
1. Interfaz y UX: Una ventana principal responsiva con un diseño moderno. Usa `ttk` y una paleta de colores profesional (grises, azules oscuros, blanco), evitando la apariencia antigua por defecto.
2. Formularios de Entrada: Un panel lateral para registrar: Código de Pieza (alfanumérico único), Nombre, Cantidad en Stock (entero positivo) y Precio Unitario (decimal positivo).
3. Validación de Datos: Implementa validaciones estrictas en las entradas para evitar que el sistema falle si se ingresan letras en campos numéricos o valores negativos. Muestra pop-ups de error descriptivos (`messagebox`).
4. Visualización: Una tabla central (`Treeview`) con barras de desplazamiento (scrollbars) que muestre los productos. Permite ordenar las columnas al hacer clic en los encabezados.
5. Controles y Búsqueda: Botones estilizados e intuitivos para: Agregar, Actualizar, Eliminar (con diálogo de confirmación de seguridad) y Limpiar campos. Incluye una barra de búsqueda en tiempo real para filtrar productos por nombre o código.
6. Persistencia y Arquitectura: Separa la lógica de datos de la interfaz. Implementa funcionalidad real para guardar y cargar desde un archivo local 'inventario.csv', asegurando que los datos se guarden automáticamente después de cada operación.
</instruction>

<format>
Proporciona el código completo, estructurado e indentado correctamente en un único archivo `app_inventario.py`. 
Debes indicar que este archivo se debe guardar dentro de una carpeta local llamada `proyectos`.
Incluye Type Hinting (tipado estático), docstrings para cada función/clase, y comentarios didácticos explicando la gestión de eventos y el uso de layouts (grid/pack).
</format>
```

## 📊 2. Python + Streamlit: Dashboard Analítico y Storytelling

**Caso de uso de la vida real:** Un Científico de Datos necesita presentar los resultados de un análisis de ventas a la junta directiva, mostrando insights clave mediante una narrativa visual clara.
**Herramienta:** Python con Streamlit y Pandas (El estándar para dashboards interactivos de Data Science).

```xml
<role>
Eres un Lead Data Scientist experto en Python, Pandas, análisis exploratorio de datos (EDA) y diseño de Dashboards con Streamlit. Tienes un talento excepcional para el Storytelling analítico, transformando datos crudos en narrativas de negocio accionables.
</role>

<context>
La junta directiva (C-level) de una cadena de retail internacional está alarmada por una caída inusual de ingresos en el último trimestre. Tienen 5 minutos para revisar tu reporte. Requieres crear un dashboard interactivo que no sea sólo un 'tablero de métricas', sino una presentación guiada que cuente la historia detrás de los datos, identifique cuellos de botella y culmine con recomendaciones claras, usando las mejores prácticas de visualización de datos.
</context>

<instruction>
Desarrolla un dashboard interactivo de alto impacto usando Python y Streamlit.
Requisitos técnicos y narrativos:
1. Generación de Datos: Escribe código Pandas para generar un dataset simulado robusto (mínimo 1000 registros) de ventas diarias del último año (fechas, regiones, categorías, ingresos, costos). Introduce intencionalmente una caída del 25% en ingresos en el último trimestre, específicamente en la región 'Norte' y categoría 'Electrónica'.
2. Optimización: Utiliza `@st.cache_data` para simular la carga eficiente de datos en un entorno real.
3. Narrativa y Layout: Estructura la página usando columnas (`st.columns`) y contenedores. Inicia con un resumen ejecutivo (Métricas principales con deltas rojo/verde usando `st.metric`). Utiliza `st.markdown` para guiar al lector paso a paso.
4. Interactividad: Añade filtros en la barra lateral (Sidebar) por rango de fechas y categorías que actualicen dinámicamente todo el dashboard.
5. Visualizaciones Avanzadas: Crea al menos 3 gráficos interactivos con Plotly o Altair: 
   - Un gráfico de líneas de tendencia (identificando la caída).
   - Un gráfico de barras comparativo (MoM o YoY).
   - Un mapa de calor o treemap para descubrir qué producto/región causó la caída.
   Aplica un esquema de colores coherente que resalte el problema (ej. gris para contexto, rojo para el problema).
</instruction>

<format>
Entregable: Un único archivo `dashboard_ventas.py` y un archivo `requirements.txt`.
Debes guardar todos los archivos generados dentro de una carpeta local llamada `proyectos`.
El código debe ser modular (funciones separadas para cargar datos, procesar y graficar). Explica brevemente qué comandos ejecutar en la terminal (`streamlit run ...`) para levantar el dashboard.
</format>
```

## ☕ 3. Java: Sistema de Gestión para Clínica Veterinaria

**Caso de uso de la vida real:** Una clínica veterinaria que necesita un sistema transaccional robusto, tipado y orientado a objetos, listo para escalar en el futuro.
**Herramienta:** Java Swing (Estructurado para copiar y pegar en NetBeans).

```xml
<role>
Eres un Arquitecto de Software Java Enterprise Senior, experto en diseño orientado a objetos, patrones de diseño (MVC, DAO) y construcción de aplicaciones de escritorio empresariales con Swing. Escribes código limpio, mantenible y escalable.
</role>

<context>
Una clínica veterinaria en expansión necesita un sistema local centralizado para gestionar el registro de pacientes. El sistema actual en Excel es insostenible. Este proyecto será la base de un sistema más grande, por lo que la arquitectura desde el día 1 debe ser perfecta. Será importado directamente por desarrolladores Junior en su IDE Apache NetBeans; la estructura de paquetes, el manejo de excepciones y el encapsulamiento deben servir como ejemplo educativo de las mejores prácticas en Java.
</context>

<instruction>
Desarrolla el código core para un Sistema de Gestión Veterinaria en Java Swing siguiendo estrictamente el patrón MVC (Modelo-Vista-Controlador).
Características:
1. Interfaz Gráfica (Vista): Crea una ventana principal que contenga un formulario completo para registrar pacientes (ID, Nombre, Especie, Raza, Edad, Dueño, Teléfono de Contacto, Síntomas). Usa márgenes profesionales (`EmptyBorder`), alineación correcta y colores neutros.
2. Tabla de Datos: Incluye un `JTable` dentro de un `JScrollPane` que se actualice dinámicamente al guardar o eliminar pacientes.
3. Modelo de Dominio: Implementa una clase `Paciente` con encapsulamiento estricto (getters/setters), constructores sobrecargados y método `toString()`.
4. Persistencia Simulada (DAO): Crea una clase que implemente un patrón DAO simple usando un `ArrayList<Paciente>` en memoria, preparando el terreno para una futura conexión a base de datos.
5. Validaciones y UX: El controlador debe manejar validaciones (ej. evitar campos vacíos, asegurar que la edad sea un número válido) usando bloques `try-catch` y mostrar mensajes `JOptionPane` para feedback de éxito o error.
6. Limpieza: Al guardar exitosamente, los campos del formulario deben limpiarse automáticamente.
</instruction>

<format>
Proporciona los archivos de código organizados lógicamente, listos para copiar y pegar en un proyecto de NetBeans. 
Debes guardar todos los archivos generados dentro de una carpeta local llamada `proyectos`.
Indica claramente la estructura de paquetes sugerida en formato de comentarios al inicio, por ejemplo:
- `com.veterinaria.modelo.Paciente.java`
- `com.veterinaria.dao.PacienteDAO.java`
- `com.veterinaria.vista.VentanaPrincipal.java`
- `com.veterinaria.controlador.PacienteController.java`
- `com.veterinaria.Main.java`
Añade comentarios JavaDoc descriptivos en las clases y métodos principales.
</format>
```

## 🏪 4. Java: Punto de Venta (POS) para Cafetería

**Caso de uso de la vida real:** Una cafetería requiere una interfaz ultra rápida para que los cajeros tomen órdenes en pantallas táctiles y calculen el total con impuestos al instante.
**Herramienta:** Java Swing (Enfoque en Layouts avanzados y manejo de eventos).

```xml
<role>
Eres un Ingeniero Java Senior experto en UI/UX de alto rendimiento y diseño de sistemas POS (Point of Sale). Tienes profundo conocimiento en el manejo avanzado de Layouts de Swing, concurrencia (Event Dispatch Thread) y ergonomía visual para pantallas táctiles.
</role>

<context>
Una concurrida cafetería de especialidad requiere un sistema POS ultrarrápido para las horas pico. Los cajeros usan pantallas táctiles y no pueden perder tiempo buscando opciones. El sistema debe ser a prueba de errores, prevenir dobles clics accidentales, recalcular totales al milisegundo y poseer una interfaz de alto contraste, moderna y táctil-friendly.
</context>

<instruction>
Diseña un Sistema POS funcional y ergonómico usando Java Swing.
Requisitos visuales, lógicos y de rendimiento:
1. Estructura de Pantalla: Divide la ventana usando `JSplitPane` (70% izquierda, 30% derecha).
   - Panel Izquierdo (Catálogo Táctil): Configurado en `GridLayout`. Muestra botones grandes (mínimo 100x100px) para productos: 'Café Americano ($35)', 'Latte ($45)', 'Capuchino ($50)', 'Muffin ($30)'. Usa colores diferenciados por categoría (bebidas vs alimentos).
   - Panel Derecho (Ticket y Cobro): Un `JTable` de solo lectura (columnas: Producto, Cantidad, Precio) y un área inferior fijada para totales.
2. Lógica de Negocio: Al hacer clic en un producto, debe añadirse al ticket. Si el producto ya existe en el ticket, incrementa la cantidad (no añadas una nueva fila). Actualiza el Subtotal, IVA (16%) y Total Neto instantáneamente.
3. Operaciones de Venta: Incluye botones grandes para 'Eliminar Ítem Seleccionado', 'Cancelar Orden' y 'COBRAR'. El botón cobrar debe resaltar (color verde) y al presionarse simular la impresión de ticket, mostrando un diálogo de éxito y reiniciando el estado.
4. Apariencia Profesional: Usa `UIManager` para configurar el Look and Feel 'Nimbus' o FlatLaf. Asegura fuentes grandes y legibles (ej. Arial 16pt) en toda la UI.
</instruction>

<format>
Entrega un solo archivo ejecutable `PuntoDeVentaCafeteria.java` que contenga el método `main` y las clases internas necesarias (Modelos, Vistas y Controladores) para facilitar su compilación rápida. 
Debes guardar este archivo dentro de una carpeta local llamada `proyectos`.
Explica mediante comentarios cómo se garantiza que las actualizaciones visuales ocurran en el Event Dispatch Thread (`SwingUtilities.invokeLater`).
</format>
```

## 🩺 5. Web Frontend: Landing Page de Consultorio Dental

**Caso de uso de la vida real:** Un grupo de dentistas necesita presencia web profesional para captar pacientes, requiriendo un diseño que transmita higiene, confianza y tecnología.
**Herramienta:** HTML5, CSS3, JavaScript Vainilla (Arquitectura Premium).

```xml
<role>
Eres un Arquitecto Frontend Senior y Lead UI/UX Designer, premiado internacionalmente por crear experiencias web inmersivas, ultra-optimizadas, accesibles (WCAG 2.1) y orientadas a la conversión. Dominas el HTML5 semántico, CSS3 moderno (Grid/Flexbox/Variables) y JavaScript Vanilla avanzado.
</role>

<context>
La 'Clínica Dental Sonrisas Blancas' es una institución premium de cirugía y estética dental. Necesitan una Landing Page que no sea una plantilla común; debe gritar "vanguardia, higiene y extrema confianza". El público objetivo son pacientes de alto poder adquisitivo que toman decisiones basadas en la primera impresión visual. El sitio debe cargar en milisegundos y estar perfectamente optimizado para SEO local y dispositivos móviles.
</context>

<instruction>
Construye una Landing Page completa (One-Page) de grado de producción.
Especificaciones de Arquitectura y Diseño:
1. Semántica y SEO: Usa etiquetas correctas (`<header>`, `<nav>`, `<main>`, `<section id="...">`, `<footer>`). Incluye meta tags básicos de SEO en el `<head>`.
2. Navegación: Navbar responsivo. En desktop, efecto cristal (Glassmorphism) fijo al hacer scroll. En móvil, un menú hamburguesa funcional con JS.
3. Hero Section: Título principal impactante (`<h1>`), subtítulo persuasivo, y un Call to Action (CTA) dual: "Agendar Cita" (botón primario) y "Conocer Especialistas" (secundario). Usa un fondo degradado premium o un patrón sutil.
4. Sección de Servicios: Cuadrícula usando CSS Grid. Tarjetas con iconografía (puedes simular con emojis o SVG base), título y breve descripción. Incluye micro-interacciones hover (levantamiento suave, sombras dinámicas, transiciones de 0.3s).
5. Social Proof y Contacto: Sección de testimonios con formato de "tarjetas de reseña". Un footer con un formulario de contacto limpio, horarios de atención y enlaces estructurados.
6. Sistema de Diseño: Implementa variables CSS (`:root`) para manejar el tema: tonos cian y azul médico, blanco puro, grises para textos. Usa 'Inter' o 'Roboto' de Google Fonts.
7. JavaScript: Implementa smooth scrolling para los enlaces del navbar, la lógica del menú móvil y un observador de intersección (Intersection Observer) para hacer "fade-in" de los elementos a medida que se hace scroll.
</instruction>

<format>
Proporciona el código listo para despliegue en tres bloques separados.
Debes guardar todos los archivos generados dentro de una carpeta local llamada `proyectos`:
1. `index.html` (Estructura semántica)
2. `styles.css` (Estilos modulares y responsivos)
3. `app.js` (Interactividad)
Añade comentarios explicando las decisiones de accesibilidad (ARIA) y la lógica del Intersection Observer.
</format>
```

## 🛍️ 6. Web Frontend: Tienda en Línea (E-commerce) de Moda

**Caso de uso de la vida real:** Una marca de ropa exclusiva busca una interfaz web (Storefront) que proyecte lujo, donde la interacción con los productos sea fluida y el diseño no compita con las fotos de la ropa.
**Herramienta:** HTML5, CSS3, JavaScript.

```xml
<role>
Eres un Ingeniero Frontend Principal y Diseñador UI/UX de clase mundial, especializado en el nicho de e-commerce de lujo y alta conversión. Tu código es impecable, modular, y dominas la creación de interfaces donde el diseño invisible hace que el producto destaque.
</role>

<context>
La marca de alta costura 'Lumière' está lanzando su nueva colección exclusiva. Su antigua web era lenta y ruidosa. Necesitan un Storefront (Catálogo) sumamente minimalista, elegante, y optimizado para móviles. La filosofía de diseño exige abundante 'whitespace' (espacios en blanco negativos), tipografía refinada y animaciones que transmitan exclusividad sin sacrificar el rendimiento web (Core Web Vitals).
</context>

<instruction>
Desarrolla el Storefront del Catálogo de Productos para 'Lumière'.
Componentes y Requisitos Técnicos:
1. Header Dinámico: Logo elegante centrado. Menú de categorías. A la derecha, un icono de carrito de compras con un badge dinámico que se actualiza por JS.
2. Hero Banner Atmosférico: Diseño edge-to-edge con el texto 'Colección Invierno 2026', tipografía Serif de gran tamaño, y un CTA minimalista (ej. botón transparente con borde fino).
3. Product Grid de Lujo: Muestra una cuadrícula asimétrica o CSS Grid perfectamente alineada con 6 productos. 
   - Tarjetas de producto: Imagen amplia, título con tipografía discreta, precio.
   - Hover Effects: Al pasar el ratón, la imagen debe hacer un sutil zoom-in (usando `transform: scale`), oscurecerse levemente y revelar un botón de 'Añadir al carrito' animado desde la parte inferior.
4. Lógica del Carrito (JS): Al hacer clic en 'Añadir al carrito', usa JavaScript para:
   - Incrementar el contador del carrito.
   - Disparar una animación de 'pop' o 'bounce' en el icono del carrito.
   - Mostrar una notificación toast temporal ('Producto añadido a tu bolsa') en la esquina inferior.
   - Usar `localStorage` para que el número del carrito persista al recargar la página.
5. Estética y Rendimiento: Usa variables CSS. Paleta: Blanco almendra, negro ónice, grises piedra. Fuentes: 'Playfair Display' (Serif para títulos) y 'Lato' o 'Montserrat' (Sans-Serif para detalles). Asegura que el código CSS prevenga el Layout Shift (CLS).
</instruction>

<format>
Proporciona la solución completa, comentada y lista para producción.
Debes guardar todos los archivos generados dentro de una carpeta local llamada `proyectos`:
1. Código de `index.html` (Incluyendo integración de Google Fonts).
2. Código de `styles.css` (Organizado por bloques lógicos).
3. Código de `main.js` (Manejo del DOM y localStorage).
</format>
```
