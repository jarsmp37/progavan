# 🚀 Guía de Inicio Rápido: Skills de AntiGravity

¡Bienvenido! Si estás empezando a usar el entorno de AntiGravity, notarás que tienes acceso a **Skills** (habilidades). Piensa en las *skills* como "sombreros de experto" que el asistente de IA se pone para ayudarte con tareas muy específicas (como ser un experto en Python, en Bases de Datos, en UI/UX, etc.).

## 🛠️ 1. ¿Cómo instalar las Skills?

Si apenas estás empezando, puedes instalar el paquete de habilidades increíbles (`awesome-skills`) directamente desde tu terminal. 

**Paso 1:** Abre tu terminal.

**Paso 2:** Ejecuta el siguiente comando:
```bash
npx -y antigravity-awesome-skills --antigravity
```

> **Nota para usuarios de Windows (PowerShell):** Si te aparece un error rojo diciendo que "la ejecución de scripts está deshabilitada", puedes forzar la instalación usando la línea de comandos clásica (cmd) con este comando:
```cmd
npx.cmd -y antigravity-awesome-skills --antigravity
```

## 🪄 2. ¿Cómo invocar las Skills?

Es sumamente sencillo. En la caja de chat donde hablas con el asistente, simplemente escribe el símbolo de la arroba **`@`**.

1. Escribe `@` y verás un menú desplegable.
2. Empieza a escribir para buscar (ej. `@python`).
3. Selecciona la skill.
4. Escribe tu petición normal. Ejemplo:
   > `@python-pro` Revisa este código y dime si estoy usando bien las listas.

El asistente leerá las reglas internas de esa *skill* y te contestará como un experto en esa materia.
